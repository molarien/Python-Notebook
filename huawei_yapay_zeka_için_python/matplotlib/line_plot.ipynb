```python
"""
lineplot: zaman içinde değişen veriler 

"""

import matplotlib.pyplot as plt

gunler = [1,2,3,4,5]
sicaklik = [22,24,23,25,27]

plt.plot(gunler, sicaklik)  # (x,y)
plt.title("Günlere Göre Sıcaklık") # title
plt.xlabel("Günler")    # x label
plt.ylabel("Sıcaklık")  # y label
plt.show() 
```

```python
import matplotlib.pyplot as plt

gunler = [1,2,3,4,5]
sicaklik = [22,24,23,25,27]

"""
marker: noktalar
"""

plt.plot(gunler, sicaklik, color = "red", linestyle = "--", marker = "o") 
plt.title("Günlere Göre Sıcaklık") 
plt.xlabel("Günler")    
plt.ylabel("Sıcaklık")  
plt.grid(True)    # ızgara
plt.show() 
```
