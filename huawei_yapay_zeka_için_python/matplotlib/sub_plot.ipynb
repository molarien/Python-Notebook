```python
"""
Birden fazla grafiği aynı anda gösterme
"""

import matplotlib.pyplot as plt

x = [1,2,3,4]
y1 = [10,20,30,40]
y2 = [40,30,20,10]

plt.subplot(1, 2, 1)  # 1 satır, 2 sütun, suan 1. grafik
plt.plot(x, y1)
plt.title("Grafik 1")

plt.subplot(1, 2, 2)
plt.plot(x, y2)
plt.title("Grafik 2")

plt.show()
```

```python
import matplotlib.pyplot as plt


x = [1,2,3,4]
y = [10,20,30,40]

plt.subplot(1, 2, 1)
plt.plot(x, y)
plt.title("Line plot")

plt.subplot(1, 2 ,2)
plt.bar(x, y)
plt.title("Bar chart")

plt.show()
```

```python
import matplotlib.pyplot as plt


x = [1,2,3,4]
y = [10,20,30,40]

plt.subplot(2 ,2, 1)
plt.plot(x, y)
plt.title("Grafik 1")

plt.subplot(2 ,2, 2)
plt.bar(x, y)
plt.title("Grafik 2")

plt.subplot(2 ,2, 3)
plt.scatter(x, y)
plt.title("Grafik 3")

plt.subplot(2 ,2, 4)
plt.pie(y)
plt.title("Grafik 4")

plt.show()
```
