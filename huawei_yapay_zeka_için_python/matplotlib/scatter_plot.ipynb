```python
"""
scatterplot: iki değişken arasındaki ilişkiyi görmek için
"""

import matplotlib.pyplot as plt

calisma_saatleri = [1,2,3,4,5,6]
notlar = [50, 55, 65, 70, 80, 90]

plt.scatter(calisma_saatleri, notlar)
plt.show()
```

```python
import matplotlib.pyplot as plt

calisma_saatleri = [1,2,3,4,5,6]
notlar = [50, 55, 65, 70, 80, 90]

# s = nokta boyutu

plt.scatter(calisma_saatleri, notlar, s = 100, color = "red")
plt.title("Çalışma Süresi ve Sınav Notları")
plt.xlabel("Çalışma Saatleri")
plt.ylabel("Notlar")
plt.show()
```

```python
import matplotlib.pyplot as plt

x1 = [1,2,3,4]
y1 = [50,60,70,80]

x2 = [1,2,3,4]
y2 = [55,65,75,85]

plt.scatter(x1, x2, color = "blue", label = "fen")
plt.scatter(x2, y2, color = "red", label = "mat")
plt.legend()   # label'ların görünür olması için
plt.show()
```
