```python
dosya = open("dosya.txt", "r", encoding = "utf-8")

for satir in dosya:
    print(satir.strip())

dosya.close()    
```

```python
dosya = open("dosya.txt", "r", encoding = "utf-8")
icerik = dosya.read()
print(icerik)
dosya.close()

new_icerik = icerik.upper()
print(f"Yeni İçerik: \n{new_icerik}")

# satır sayısını bulma

dosya = open("dosya.txt", "r", encoding = "utf-8")
satirlar = dosya.readlines()
dosya.close()

print(f"Toplam Satır: {len(satirlar)}")
```

```python
dosya = open("dosya.txt", "r", encoding = "utf-8")
icerik = dosya.read()
dosya.close()

new_icerik = icerik.upper()

dosya = open("islenmis_dosya", "w", encoding = "utf-8")
dosya.write(new_icerik)
dosya.close()
```
