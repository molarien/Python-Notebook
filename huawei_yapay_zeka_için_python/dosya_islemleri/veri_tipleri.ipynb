```python
# list,     tuple,       set
#      değiştirilemez  unıque



print(0.1 + 0.2)
print(round(0.1 + 0.2, 2))

print(str(5) + "hello")

string = "python"
print(string.replace("o", "0"))

# tuple metotları
t = (20, 20, 30, 40)
print(t.count(20))
print(t.index(30))
```

```python
# sözlük (dictionary)
ogrenci = { # isim = anahtar, ali = key değeri -> {anahtar: değer}
    "isim": "ali", 
    "yas": 25,
    "bolum": "bilgisayar"
}

del ogrenci["bolum"]
print(ogrenci)

print(ogrenci.keys())
print(ogrenci.values())
print(ogrenci.items())
```

```python
# set
sayilar = {1, 2, 3, 4}
print(sayilar)


# listeyi set e çevirme
liste = [1, 2, 2, 3, 4, 4]
benzersiz = set(liste)
print(benzersiz)


# set işlemleri
a = {1, 2, 3}
b = {3, 4, 5}

print(a.union(b)) # birleşim {1, 2, 3, 4, 5}
print(a.intersection(b)) # kesişim {3}
print(a.difference(b)) # fark {1, 2}
```
