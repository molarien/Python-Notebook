```python
"""
init
"""

class Ogrenci:

    def __init__(self, isim, yas):    # self = oluşturulan nesneyi temsil eder
        print(f"Yeni bir öğrenci oluşturuluyor. İsim: {isim}, Yaş: {yas}")


ogrenci = Ogrenci("Ali", 21)
```

```python
"""
Attribute
"""

class Ogrenci:

    def __init__(self, isim, yas):
        self.isim = isim
        self.yas = yas

ogrenci = Ogrenci("Ali", 24)

print(ogrenci.isim)
print(ogrenci.yas)
```

```python
"""
method 
"""

class Ogrenci:

    def __init__(self, isim, yas):
        self.isim = isim
        self.yas = yas

    def tanit(self):
        print("Merhaba")    

ogrenci = Ogrenci("Ali", 24)
ogrenci.tanit()
```

```python
class Kitap:

    def __init__(self, ad, yazar, sayfa):
        self.ad = ad
        self.yazar = yazar
        self.sayfa = sayfa

    def bilgi(self):
        print(self.ad) 
        print(self.yazar) 
        print(self.sayfa)

kitap = Kitap("Python Prog.", "Kaan", 500)

print(kitap.ad)
print(kitap.yazar)
print(kitap.sayfa)

kitap.bilgi()
```
