```python
"""
veri seçme ve filtreleme
"""

import pandas as pd

veri = {
    "isim" : ["ali", "ayse", "mehmet", "zeynep", "ahmet"],
    "yas" : [25, 30, 40, 15, 22],
    "sehir" : ["Ankara", "İstanbul", "İzmir", "Ankara", "Bursa"],
    "maas" : [5000, 7000, 8000, 9000, 8000]
}

df = pd.DataFrame(veri)

print(df)
```

```python
# Sütun seçme

print(df["isim"])   # serie döndürür

print(df[["isim", "maas"]])
```

```python
# Satır Seçme

print(df.iloc[0])

# Birden fazla satır seçme

print(df.iloc[0:3])

# Satır seçme (loc)

print(df.loc[2])      

"""
.iloc[] -> indeks değerine göre seçer
.loc[]  -> etiketine göre seçer (key) 

"""
```

```python
# Belirli bir satır ve belirli bir sütun

print(df.loc[:, ["isim", "maas"]])
```

```python
# Koşullu filtreleme

filtre = df["yas"] > 30

print(filtre)


"""
0    False
1    False
2     True
3    False
4    False
"""

result = df[filtre]
print(result)
```

```python
print(df[df["yas"] > 30])
```

```python
# Birden fazla koşul varsa

sonuc = df[(df["sehir"] == "Ankara") & (df["maas"] > 6000)]

print(sonuc)
```

```python
# Belirli bir değeri içeren satırlar

print(df[df["sehir"] == "Ankara"])
```

```python
# Sadece belirli sütunları gösterme

print(df[df["yas"] > 20][["isim", "yas"]])
```
