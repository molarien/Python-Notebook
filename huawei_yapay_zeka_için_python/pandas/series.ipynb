```python
import pandas as pd

veri = pd.Series([10,20,30,40])

print(veri)


"""
index  value
0       10
1       20
2       30
3       40

key-value olarak tutulur ex: 1:20, 2:30
"""

print(veri[0])
print(veri[1])
```

```python
veri = pd.Series([10,20,30], index = ["a", "b", "c"])

print(veri)

"""

a    10
b    20
c    30


"""

print(veri["b"])
```

```python
veri = {
    "ali" : 80,
    "ayşe" : 90,
    "mehmet" : 75
}

s = pd.Series(veri)

print(s)

print(s.index)
print(s.values)
print(s.dtype)
```

```python
veri = pd.Series([10,20,30,40])

result = veri * 2

print(result)
```

```python
age = pd.Series([10,20,30,40,50])

filtre = age > 25                #0    False
                                 #1    False
                                 #2     True
                                 #3     True
                                 #4     True
print(filtre)

result = age[filtre]             #2    30
                                 #3    40
                                 #4    50
print(result)
```
