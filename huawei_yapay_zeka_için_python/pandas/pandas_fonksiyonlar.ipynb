```python
import pandas as pd

veri = {
    "isim" : ["ali", "ayşe", "mehmet"],
    "yas" : [25,30,28],
    "sehir" : ["Ankara", "İstanbul", "İzmir"]
}

df = pd.DataFrame(veri)    

print(df)
```

```python
# ilk X satırı görme

print(df.head(10))
```

```python
# son X satırı görme

print(df.tail(2))
```

```python
print(df.info())
```

```python
# sayısal sütunların temel istatistiklerini görmek için describe()


print(df.describe())
```

```python
# bir sütundaki değerlerin kaç kez tekrar ettiğini görmek için value_count() kullanılır

print(df["sehir"].value_counts())
```

```python
# bir sütundaki benzersiz değerleri görmek için unique() kullanırız

print(df["sehir"].unique())
```

```python
# bir sütunda kaç farklı değer old görmek için

print(df["sehir"].nunique())
```
