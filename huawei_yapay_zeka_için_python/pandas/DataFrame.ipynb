```python
"""
dataframe
"""

import pandas as pd
 
veri = {
    "isim" : ["ali", "ayşe", "mehmet"],
    "yas" : [25,30,28],
    "sehir" : ["Ankara", "İstanbul", "İzmir"]
}

df = pd.DataFrame(veri)    

"""
     isim  yas     sehir
0     ali   25    Ankara
1    ayşe   30  İstanbul
2  mehmet   28     İzmir
"""

print(df)
```

```python
# Sütun isimleri

print(df.columns)
```

```python
# satır sayısı 

print(df.shape)   # 3 satır, 3 sütun
```

```python
# sutünlara erişim

print(df["isim"])
```

```python
# birden fazla sütun seçme

print(df[["isim", "yas"]])
```

```python
# yeni sütun ekleme

df["maas"] = [5000, 7000, 6000]

print(df)
```

```python
# sütun silme

df = df.drop("sehir", axis = 1)         # axis = 0    satır
                                        # axis = 1    sütun

print(df)
```
