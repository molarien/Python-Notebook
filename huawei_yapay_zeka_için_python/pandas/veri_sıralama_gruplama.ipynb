```python
import pandas as pd


veri = {
    "isim" : ["ali", "ayse", "mehmet", "zeynep", "ahmet"],
    "yas" : [25, 30, 40, 15, 22],
    "sehir" : ["Ankara", "İstanbul", "İzmir", "Ankara", "Bursa"],
    "maas" : [5000, 7000, 8000, 9000, 8000]
}

df = pd.DataFrame(veri)

print(df)
```

```python
# veri sıralama

df_sirali = df.sort_values("maas")

print(df)
```

```python
# azalan sıralama

df_sirali = df.sort_values("maas", ascending = False)

print(df_sirali)
```

```python
# birden fazla sütuna göre sıralama

df_sirali = df.sort_values(["sehir", "maas"])

print(df_sirali)
```

```python
# veri gruplama  (groupby)

gruplar = df.groupby("sehir")

print(gruplar)     # <pandas.api.typing.DataFrameGroupBy object at 0x000001F20E4FC1A0>

```

```python
sonuc = df.groupby("sehir")["maas"].mean()       # şehir bazında ort maaş hesaplama

print(sonuc)
```

```python
# grupların toplam maaşı

sonuc = df.groupby("sehir")["maas"].sum()

print(sonuc)
```

```python
# grupların kaç kişi olduğunu bulma

sonuc = df.groupby("sehir")["isim"].count()

print(sonuc)
```

```python
# birden fazla işlem yapma

sonuc = df.groupby("sehir")["maas"].agg(["mean", "max", "min"])

print(sonuc)
```
