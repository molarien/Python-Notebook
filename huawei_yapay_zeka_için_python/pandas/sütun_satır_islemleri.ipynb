```python
import pandas as pd 

veri = {
    "isim" : ["ali", "ayse", "mehmet"],
    "yas" : [25,30,28],
    "maas" : [5000, 7000, 6000]
}

df = pd.DataFrame(veri)

print(df)


# yeni sütun ekleme

df["sehir"] = ["Ankara", "İzmir", "İstanbul"] 


print(df)

# hesaplama ile sütun oluşturma

df["yillik_maas"] = df["maas"] * 12

print(df)


# sütun silme

df = df.drop("maas", axis = 1)

print(df)


# sütun isim değiştirme

df = df.rename(columns = {"yillik_maas" : "yillikMaas"})

print(df)
```

```python
# yeni satır ekleme

df.loc[3] = ["Zeynep", 32, "Ankara", 8000]

print(df)
```

```python
# satır silme

df = df.drop(0)

print(df)
```

```python
# index değerlerini yeniden düzenleme

df = df.reset_index(drop = True)

print(df)
```
