```python
try:
    num = int(input("Sayı Girin: "))
    result = 10/num

except (ValueError, ZeroDivisionError):
    print("Error")

else:
    print(f"Result: {result}")
```

```python
try:
    dosya = open("data.txt", "r", encoding = "utf-8")
    icerik = dosya.read()
    print(icerik)

except FileNotFoundError:
    print("Dosya Bulunamadı")

finally:
    try:
        dosya.close()
    except:
        pass
```
