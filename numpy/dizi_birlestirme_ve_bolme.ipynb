```python
import numpy as np

a = np.array(
    [
        [1,2],
        [3,4]
    ]
)

b = np.array(
    [
        [5,6],
        [7,8]
    ]
)

result = np.concatenate((a,b), axis = 0)    # satır yönü birleştirme

print(result)
```

```python
import numpy as np

a = np.array(
    [
        [1,2],
        [3,4]
    ]
)

b = np.array(
    [
        [5,6],
        [7,8]
    ]
)

result = np.vstack((a,b))     # vertical stack dikey birleştirme
                              # axis = 0 gibi 

print(result)
```

```python
import numpy as np

a = np.array(
    [
        [1,2],
        [3,4]
    ]
)

b = np.array(
    [
        [5,6],
        [7,8]
    ]
)

result = np.hstack((a,b))     # horizontal stack yatay birleştirme
                              # axis = 1 gibi 

print(result)
```
