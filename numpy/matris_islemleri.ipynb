```python
import numpy as np

a = np.array([
    [1,2],
    [3,4]
])

b = np.array([
    [5,6],
    [7,8]
])

print(a)
print(b)

print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

```python
# dot product

a = np.array([
    [1,2],
    [3,4]
])

b = np.array([
    [5,6],
    [7,8]
])


result = np.dot(a,b)

print(result)
```

```python
# matris transpose

a = np.array([
    [1,2],
    [3,4]
])

print(a.T)     # a.T
```

```python
# matris determinant

a = np.array([
    [1,2],
    [3,4]
])

det = np.linalg.det(a)

print(det)
```

```python
# matris inverse

a = np.array([
    [1,2],
    [3,4]
])

inv = np.linalg.inv(a)

print(inv)
```
