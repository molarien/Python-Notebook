```python
"""
numpy:  Yüksek performanslı sayısal hesaplar için
        büyük veri, hızlı mat, matris hesaplama
        numpy C dili ile yazılmış
        
"""

import numpy as np

"""
Diziler(Array)
    ndarray: n-dimensional array
"""

nums = [1,2,3,4,5]
print(nums)

nums_numpy = np.array(nums)
print(nums_numpy)

print(type(nums_numpy))

print(nums_numpy.shape)   # (5,) -> 5 elemanlı tek boyutlu dizi

print(nums_numpy.dtype)   # int64 -> integer
# numpy dizileri içerisinde tek çeşit veri tipi depolar
```

```python
dizi = np.zeros(5)

print(dizi)
```

```python
dizi = np.ones(5)

print(dizi)
```

```python
dizi = np.arange(0,10)

print(dizi)
```

```python
dizi = np.arange(0, 10, 2)

print(dizi)
```

```python
dizi = np.linspace(0, 10, 5)    # [ 0.  2.5   5.  7.5  10. ]  0 ile 10 arasında 5 sayı üret

print(dizi)
```
