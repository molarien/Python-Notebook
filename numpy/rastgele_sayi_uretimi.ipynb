```python
import numpy as np

random = np.random.rand(5)   # 0-1 arasında 5 sayı oluşturma

print(random)
```

```python
random = np.random.rand(3,3)   # 3x3 matris üretme

print(random)
```

```python
random = np.random.randint(1,10,5)  # 1-10 arasında 5 tane tamsayı üret

print(random)
```

```python
random = np.random.randint(1,20,(3,4))  # 1-20 arasında 3x4'lük matris üret

print(random)
```

```python
# aynı rastgele sonucu üretmek için

np.random.seed(42)    # bu atamayı yaptıktan sonra aynı inputa hep aynı outputu vericek
                      # seed'i değiştirdiğinde sonuçlar değişir ama 

random = np.random.rand(5)

print(random)
```

```python
dizi = np.array([10,20,30,40,50])

choice = np.random.choice(dizi)      # diziden rasgele eleman seçer

print(choice)
```

```python
dizi = np.array([10,20,30,40,50])

choice = np.random.choice(dizi, 3)    # diziden 3 tane elemanı seçer

print(choice)
```
